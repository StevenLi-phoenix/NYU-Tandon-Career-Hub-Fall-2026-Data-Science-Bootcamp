# Part 0 · Working in a Notebook

---

A notebook is a sequence of **cells**:

- **Text cells**, like this one, contain explanation written in Markdown.
- **Code cells** contain Python. Select one and press **`Shift + Enter`** to run it; the output appears directly beneath.

All cells share a single running Python process, the **kernel**. A variable defined in one cell is available in every cell executed after it. A `NameError` almost always means a cell above was skipped, so run cells in order from the top.

Any text following `#` on a line is a **comment**, which Python ignores.

````python
# This is a comment. Python ignores it.
print("Hello, world!")        # print() writes a value to the output
````

### Why Python?

**Readability.** Python is a *high-level* language: its syntax is compact and close to plain English, which makes code easier to write, review, and maintain.

**Productivity.** Python typically needs far less code than languages such as Java to accomplish the same task. Printing every line of a file in Java:

```java
import java.io.BufferedReader;
import java.io.FileReader;
import java.io.IOException;

public class CsvPrinter {
    public static void main(String[] args) {
        try (BufferedReader br = new BufferedReader(new FileReader("responses.csv"))) {
            String line;
            while ((line = br.readLine()) != null) {
                System.out.println(line);
            }
        } catch (IOException e) {
            e.printStackTrace();
        }
    }
}
```

The equivalent in Python is a single line: `print(open("responses.csv").read())`.

**The trade-off.** Pure Python executes more slowly than Java or C. For most analytical work, however, **development time dominates execution time**, and the numerical libraries you will meet next week (NumPy, pandas) delegate their heavy computation to compiled C code.

**Ecosystem.** Python ships with an extensive standard library, and its package ecosystem covers essentially every area of data work: analysis, visualization, machine learning, web services, and more.

### Loading the data

**Upload `responses.csv` first.** In Colab, open the **Files** panel in the left sidebar and drag the file in. In Jupyter, place it in the same folder as this notebook.

Then run the cell below. It reads the file and stores the result in a variable named `apps`. You need not follow its details yet; we will return to it at the end of the session, when every line will be familiar.

````python
# @title Setup: load the responses (run this cell first)
# This cell uses several ideas we have not covered yet. By the end of the session, every line will be familiar.
import csv

DATA_FILE = "responses.csv"   # in Colab, upload this file through the Files panel on the left

# The column names in the file are terse, so we map each one to a short, descriptive nickname.
NICKNAMES = {
    "Timestamp":           "timestamp",
    "Level":               "level",           # self-rated programming level
    "Major":               "major",
    "Class Level":         "class_level",
    "Lang":                "languages",       # languages used before (free text)
    "Section":             "section",
    "Experience":          "experience",      # where they have used these skills (select all that apply)
    "Interest":            "essay_excited",   # what are you most excited to learn about?
    "Gain":                "essay_gain",      # what do you hope to gain?
    "Additional Interest": "additional",      # anything else? (optional)
    "Time":                "hours",           # hours per week they can commit
}
COLUMNS = {nick: column for column, nick in NICKNAMES.items()}   # nickname -> original column name

apps = []    # a list of dictionaries: one dictionary per application
try:
    with open(DATA_FILE, newline="", encoding="utf-8") as f:
        for row in csv.DictReader(f):
            app = {}
            for column, answer in row.items():
                app[NICKNAMES[column]] = answer
            apps.append(app)
    print(f"Loaded {len(apps)} applications from {DATA_FILE}.")
except FileNotFoundError:
    print(f"Could not find {DATA_FILE}. Upload it through the Files panel on the left, then run this cell again.")
````

# Part 1 · Variables and Primitive Types

---

A **variable** is a name bound to a value. The assignment operator `=` creates the binding:

```python
major = "Computer Science"
```

Read this as "`major` *refers to* the string `'Computer Science'`" rather than "equals." Names may contain letters, digits, and underscores, may not begin with a digit, and are case-sensitive.

Python's primitive (single-value) types:

| Type | Description | Examples | In our data |
|---|---|---|---|
| `int` | Integers | `1`, `2`, `-3` | the number of applications |
| `float` | Floating-point numbers | `3.14`, `7.5` | 7.5 hours per week |
| `str` | Text | `"Hello"` | `"Computer Science"` |
| `bool` | Truth values | `True`, `False` | whether a question was skipped |
| `NoneType` | The absence of a value | `None` | an answer that does not exist |

The built-in `type()` reports the type of any value:

````python
major = "Computer Science"      # str
section = 2                     # int
max_hours = 7.5                 # float
is_beginner = False             # bool
github = None                   # NoneType

print("major       ", type(major))
print("section     ", type(section))
print("max_hours   ", type(max_hours))
print("is_beginner ", type(is_beginner))
print("github      ", type(github))
````

Python is **dynamically typed**: the type belongs to the value, not to the name. A variable can be rebound to a value of a different type at any time.

````python
section = 2
print(section, type(section))

section = "Section 2: Thursdays"     # same name, new value, new type
print(section, type(section))
````

# Part 2 · Operators and Expressions

---

An **expression** combines values and operators and evaluates to a single value, which itself has a type.

| Arithmetic | Comparison | Boolean | Membership |
|---|---|---|---|
| `+` addition | `==` equal | `not` negation | `in` |
| `-` subtraction | `!=` not equal | `and` conjunction | `not in` |
| `*` multiplication | `>` greater than | `or` disjunction | |
| `/` true division (always `float`) | `>=` greater or equal | | |
| `//` floor division | `<` less than | | |
| `%` remainder (modulo) | `<=` less or equal | | |
| `**` exponentiation | | | |

Note the distinction between `=` (assignment: *bind* a name) and `==` (comparison: *test* equality).

````python
print(7 + 2, 7 - 2, 7 * 2)
print(7 / 2)       # 3.5  true division always returns a float
print(7 // 2)      # 3    floor division rounds toward negative infinity
print(7 % 2)       # 1    remainder; useful for parity checks
print(2 ** 10)     # 1024
````

Arithmetic on real answers: each applicant selected an hours *range*. Taking the lower bound of each range as a number lets us average the first four applicants. (Part 6 shows how to extract that number programmatically; for now we transcribe it.)

````python
print(apps[0]["hours"], "|", apps[1]["hours"], "|",
      apps[2]["hours"], "|", apps[3]["hours"])
````

````python
# the lower bound of each range printed above ("5 - 7.5 hours" -> 5)
h0, h1, h2, h3 = 5, 5, 7.5, 5

total = h0 + h1 + h2 + h3
avg = total / 4
avg_floor = total // 4

print("Total:", total)
print("Mean:", avg)
print("Floor of mean:", avg_floor)
print("Mean of at least 5 hours?", avg >= 5)
````

**Comparison** operators produce booleans, and **boolean** operators combine them. Change `hour` below and predict each result before running.

````python
hour = 23    # try 2, 7, 14, 23

is_night_owl = hour >= 22 or hour < 5
is_early_bird = 5 <= hour < 9          # comparisons can be chained
is_neither = not (is_night_owl or is_early_bird)

print("Night owl:  ", is_night_owl)
print("Early bird: ", is_early_bird)
print("Neither:    ", is_neither)
````

The **membership** operator `in` tests containment. Its meaning depends on the container: for a string it tests for a substring, for a dictionary it tests the *keys*.

````python
print("Science" in "Computer Science")      # substring test
print("science" in "Computer Science")      # case-sensitive
print("hours" in first)                     # is "hours" a key of this dictionary?
print("github" in first)                    # GitHub links were removed for privacy
````

### `while` loops

A `while` loop repeats its body **as long as** its condition holds. It is the right tool when the number of iterations is not known in advance.

```python
while condition:
    ...
```

> **Question 1. How many people applied before the final day?**
>
> The form closed at the end of **9/13**. Rows are stored in submission order, so we can walk down the list until we reach the first application dated 9/13. We do not know in advance how many steps that will take, which is precisely the situation `while` is designed for.

````python
i = 0
while not apps[i]["timestamp"].startswith("9/13"):
    i += 1

print("The first 9/13 application is at index", i)
print("So", i, "people applied before the final day")
````

````python
# Equivalent formulation using break
i = 0
while True:                  # loop indefinitely...
    if apps[i]["timestamp"].startswith("9/13"):
        break                # ...until break exits the loop
    i += 1

print("Same result with break:", i)
before_last_day = i
````

### `for` loops

A `for` loop executes its body **once for each element** of an iterable: a list, a string, a range, the keys of a dictionary, and so on.

```python
for item in iterable:
    ...
```

````python
for lang in ["Python", "Java", "SQL"]:
    print("Language:", lang)

for character in "Hi!":
    print(character)
````

Now count **every** application submitted on the final day:

````python
last_day = 0
for app in apps:
    if app["timestamp"].startswith("9/13"):
        last_day += 1

print("Before the final day:", before_last_day)
print("On the final day:    ", last_day)
print("Accounted for:       ", before_last_day + last_day, "of", len(apps))
````

> **Finding.** The totals do not reconcile. **Several applications arrived after the form officially closed.** We will examine those rows in Part 5.

> **Question 1b. Who skipped questions?**

Truthiness gives us a concise test for blank answers:

````python
skipped_languages = 0
skipped_additional = 0

for app in apps:
    if not app["languages"]:          # an empty string is falsy, so `not` makes the test true
        skipped_languages += 1
    if not app["additional"]:
        skipped_additional += 1

print("Left 'languages' blank: ", skipped_languages)
print("Left 'additional' blank:", skipped_additional)
````

Loops can be **nested**. Here we audit every column at once:

````python
for question in COLUMNS:              # iterating over a dictionary yields its keys
    blanks = 0
    for app in apps:
        if app[question].strip() == "":
            blanks += 1
    print(question.rjust(14), "blank answers:", blanks)
````

**Principle.** `len()` tells you how many rows exist, not what they contain. Always inspect the data before trusting a summary statistic.

### The central pattern of EDA: counting with a dictionary

> **Question 2. Who applied, and what do they study?**

Begin with an empty dictionary. For each application, inspect the answer: if it has been seen before, increment its count; otherwise, initialize its count to 1.

````python
major_counts = {}

for app in apps:
    major = app["major"]
    if major in major_counts:
        major_counts[major] += 1
    else:
        major_counts[major] = 1

print(major_counts)
````

A raw dictionary is difficult to read. A **text bar chart** is easy to build from two string operations: repetition (`"█" * 3` produces `"███"`) and `.rjust()`, which right-aligns text within a fixed width.

````python
print("█" * 3)

for major, count in major_counts.items():
    print(major.rjust(38), "█" * count, count)
````

The `if`/`else` counting pattern has a compact idiom using `.get(key, default)`:

````python
level_counts = {}
for app in apps:
    level = app["level"]
    level_counts[level] = level_counts.get(level, 0) + 1    # current count (or 0 if unseen) plus one

print(level_counts)
````

````python
class_counts = {}
for app in apps:
    cl = app["class_level"]
    class_counts[cl] = class_counts.get(cl, 0) + 1

for cl, count in class_counts.items():
    print(cl.rjust(20), "█" * count, count)
````

Strings, like tuples, are **immutable**. Every string method returns a *new* string; the original is never modified in place.

#### `.split()`

Slicing by position is brittle. `.split()` divides a string at every occurrence of a separator, regardless of how long the pieces are:

````python
hour_counts = {}

for app in apps:
    pieces = app["timestamp"].split(" ")     # ['8/26/2026', '14:48:44'] or ['9/15/26', '3:30', 'PM']
    hour = int(pieces[1].split(":")[0])
    if len(pieces) == 3 and pieces[2] == "PM" and hour != 12:
        hour += 12                             # 3 PM -> 15
    elif len(pieces) == 3 and pieces[2] == "AM" and hour == 12:
        hour = 0                               # 12 AM -> 0
    hour_counts[hour] = hour_counts.get(hour, 0) + 1

print(sum(hour_counts.values()), "timestamps counted")   # sanity check: should equal len(apps)

# print hours in clock order, including hours with no submissions
for hour in range(24):
    count = hour_counts.get(hour, 0)
    print(str(hour).rjust(2) + ":00", "█" * count, count)
````

### A small analytics toolkit

Count how many times we have written the same counting loop: majors, levels, class levels, hours, dates. Repetition like this is sometimes called **WET** code (*write everything twice*). The remedy is the **DRY** principle (*don't repeat yourself*): factor the pattern into reusable functions.

We also introduce **default parameter values**. In `top=None`, the parameter takes the value `None` unless the caller supplies another, typically as a **keyword argument** such as `top=10`.

````python
def count_values(rows, field):
    """Count how many times each answer appears in one column."""
    counts = {}
    for row in rows:
        answer = row[field].strip()
        if answer == "":
            answer = "(no answer)"
        counts[answer] = counts.get(answer, 0) + 1
    return counts


def bar_chart(counts, title="", top=None, width=24):
    """Print a text bar chart from a {label: count} dictionary, largest first."""
    # 1. convert the dictionary into (count, label) pairs, largest first
    pairs = []
    for label, count in counts.items():
        pairs.append((count, label))
    pairs.sort(reverse=True)

    # 2. keep only the top N, if requested
    if top is not None:
        pairs = pairs[:top]

    # 3. print a title, then one bar per label, scaled so the longest bar is 40 blocks
    if title != "":
        print(title)
        print("=" * 60)
    biggest = pairs[0][0]
    for count, label in pairs:
        bar_length = max(1, count * 40 // biggest)   # at least one block, so small counts remain visible
        print(str(label)[:width].rjust(width), "│" + "█" * bar_length, count)
    print()
````

The rows from the case study reappear: three different spellings of `5 - 7.5 hours`.

We will write **cleaning functions** and store the cleaned values under **new keys** in each application. Deriving new variables from existing ones is called **feature engineering**.

````python
def clean_hours(answer):
    """Normalize '5-7.5 hours' and '5 – 7.5 hours' to '5 - 7.5 hours'."""
    answer = answer.replace("–", "-")          # en dash -> hyphen
    answer = answer.replace(" - ", "-")        # remove spaces around the hyphen...
    answer = answer.replace("-", " - ")        # ...then restore them consistently
    return answer.strip()


def clean_class_level(answer):
    """Correct the misspelling that appears in the form itself."""
    return answer.replace("Sophmore", "Sophomore").strip()


def degree_type(class_level):
    """Group detailed class levels into three categories."""
    cl = class_level.lower()
    if "masters" in cl:
        return "Master's"
    elif "phd" in cl:
        return "PhD"
    else:
        return "Undergraduate"


# test functions before trusting them
print(clean_hours("5-7.5 hours"), "|", clean_hours("5 – 7.5 hours"), "|", clean_hours("5 - 7.5 hours"))
print(clean_class_level("Sophmore"))
print(degree_type("Second year masters"))
````

````python
for app in apps:
    app["hours_clean"] = clean_hours(app["hours"])
    app["class_clean"] = clean_class_level(app["class_level"])
    app["degree"] = degree_type(app["class_clean"])

bar_chart(count_values(apps, "hours_clean"), "Weekly time commitment (cleaned)")
bar_chart(count_values(apps, "class_clean"), "Class level (cleaned)")
bar_chart(count_values(apps, "degree"), "Degree type")
````

### Handling errors with `try` / `except`

Some operations can fail on unexpected input. Rather than letting the program halt, we can **catch** the exception and respond deliberately. `float()`, for instance, raises `ValueError` on text that is not numeric:

````python
def to_number(text):
    """Convert text to a float, returning None if it is not numeric."""
    try:
        return float(text)
    except ValueError:            # executes only if float() raised ValueError
        return None

print(to_number("7.5"))
print(to_number("seven"))
````

2. **Local scope:** names created *inside* a function, including its parameters, exist only while that call is executing and are invisible outside it.

The last line of the next cell raises an exception by design.

````python
DEADLINE = "9/13/2026"            # global: visible everywhere

def count_on_deadline(rows):
    on_deadline = 0               # local: exists only inside this function
    for row in rows:
        if row["timestamp"].startswith(DEADLINE):   # reading a global is permitted
            on_deadline += 1
    return on_deadline

print("Submitted on the deadline day:", count_on_deadline(apps))
print("DEADLINE is still:", DEADLINE)
print(on_deadline)                # local names do not survive the call
````

````python
def add_bonus_hours(hours):
    hours = hours + 2             # rebinds the LOCAL name only
    print("Inside the function:", hours)
    return hours

my_hours = 5
print("Returned value:", add_bonus_hours(my_hours))
print("my_hours outside is unchanged:", my_hours)
````

**Mutable arguments behave differently.** Recall aliasing from Part 5: passing a list or dictionary to a function passes a reference to the *same* object. If the function *mutates* that object (for example with `.append()`), the change is visible to the caller. This is exactly how our feature-engineering loops added new keys to every application.

````python
def add_language(lang_list, lang):
    lang_list.append(lang)        # mutates the caller's list

my_langs = ["Python"]
add_language(my_langs, "SQL")
print(my_langs)                   # ['Python', 'SQL']
````

````python
def top_item(counts):
    """Return the label with the highest count."""
    best_label, best_count = None, -1
    for label, count in counts.items():
        if count > best_count:
            best_label, best_count = label, count
    return best_label


def line(label, value):
    print("  " + label.ljust(34, ".") + " " + str(value))


print("╔" + "═" * 58 + "╗")
print("   DATA SCIENCE BOOTCAMP · FALL 2026 · APPLICANT SUMMARY")
print("╚" + "═" * 58 + "╝")
line("Applications", len(apps))
line("Applied before the final day", before_last_day)
line("Applied on the final day (9/13)", last_day)
line("Applied after the deadline", len(apps) - before_last_day - count_on_deadline(apps))
line("Most common major", top_item(count_values(apps, "major")))
line("Most common class level", top_item(count_values(apps, "class_clean")))
line("Most popular section", top_item(count_values(apps, "section")))
line("Most common time commitment", top_item(count_values(apps, "hours_clean")))
line("Most widely known language", top_item(language_counts))
line("Night-owl submissions (10PM-5AM)", night_owls)
line("Most anticipated topic", top_item(topic_counts))
line("Top essay word (after cleaning)", top_item(word_counts))
````
