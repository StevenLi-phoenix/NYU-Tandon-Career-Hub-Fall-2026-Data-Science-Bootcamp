### Why Python?

**Readability.** Python is a *high-level* language: its syntax is compact and close to plain English, which makes code easier to write, review, and maintain.

**Productivity.** Python typically needs far less code than languages such as Java to accomplish the same task. Printing every line of a file in Java:

```java
import java.io.BufferedReader;
import java.io.FileReader;
import java.io.IOException;

public class CsvPrinter {
    public static void main(String[] args) {
        try (BufferedReader br = new BufferedReader(new FileReader("responses.csv"))) {
            String line;
            while ((line = br.readLine()) != null) {
                System.out.println(line);
            }
        } catch (IOException e) {
            e.printStackTrace();
        }
    }
}
```

The equivalent in Python is a single line: `print(open("responses.csv").read())`.

**The trade-off.** Pure Python executes more slowly than Java or C. For most analytical work, however, **development time dominates execution time**, and the numerical libraries you will meet next week (NumPy, pandas) delegate their heavy computation to compiled C code.

**Ecosystem.** Python ships with an extensive standard library, and its package ecosystem covers essentially every area of data work: analysis, visualization, machine learning, web services, and more.

A **variable** is a name bound to a value. The assignment operator `=` creates the binding:

```python

```

Read this as "`major` *refers to* the string `'Computer Science'`". Names may contain letters, digits, and underscores, may not begin with a digit, and are case-sensitive.

| Type | Description | Examples | In our data |

````python
# @title Setup: load the responses (run this cell first)
# This cell uses several ideas we have not covered yet. By the end of the session, every line will be familiar.
import csv

DATA_FILE = "responses.csv"   # in Colab, upload this file through the Files panel on the left

# The column names in the file are terse, so we map each one to a short, descriptive nickname.
NICKNAMES = {
    "Timestamp":           "timestamp",
    "Level":               "level",           # self-rated programming level
    "Major":               "major",
    "Class Level":         "class_level",
    "Lang":                "languages",       # languages used before (free text)
    "Section":             "section",
    "Experience":          "experience",      # where they have used these skills (select all that apply)
    "Interest":            "essay_excited",   # what are you most excited to learn about?
    "Gain":                "essay_gain",      # what do you hope to gain?
    "Additional Interest": "additional",      # anything else? (optional)
    "Time":                "hours",           # hours per week they can commit
}
COLUMNS = {nick: column for column, nick in NICKNAMES.items()}   # nickname -> original column name

apps = []    # a list of dictionaries: one dictionary per application
try:
    with open(DATA_FILE, newline="", encoding="utf-8") as f:
        for row in csv.DictReader(f):
            app = {}
            for column, answer in row.items():
                app[NICKNAMES[column]] = answer
            apps.append(app)
    print(f"Loaded {len(apps)} applications from {DATA_FILE}.")
except FileNotFoundError:
    print(f"Could not find {DATA_FILE}. Upload it through the Files panel on the left, then run this cell again.")
````

# Part 2 · Operators and Expressions

---

An **expression** combines values and operators and evaluates to a single value, which itself has a type.

| Arithmetic | Comparison | Boolean | Membership |
|---|---|---|---|
| `+` addition | `==` equal | `not` negation | `in` |
| `-` subtraction | `!=` not equal | `and` conjunction | `not in` |
| `*` multiplication | `>` greater than | `or` disjunction | |
| `/` true division (always `float`) | `>=` greater or equal | | |
| `//` floor division | `<` less than | | |
| `%` remainder (modulo) | `<=` less or equal | | |
| `**` exponentiation | | | |

Note the distinction between `=` (assignment: *bind* a name) and `==` (comparison: *test* equality).

````python
print(7 + 2, 7 - 2, 7 * 2)
print(7 / 2)       # 3.5  true division always returns a float
print(7 // 2)      # 3    floor division rounds toward negative infinity
print(7 % 2)       # 1    remainder; useful for parity checks
print(2 ** 10)     # 1024
````

#### Indexing and slicing

Indices start at **0**; negative indices count from the end. A slice `start:stop` includes `start` and excludes `stop`, so `len(x[a:b]) == b - a`.

````python
nums = list(range(5))   # range(5) yields 0, 1, 2, 3, 4
print(nums)
print(nums[0])           # first element
print(nums[-1])          # last element
print(nums[2:4])         # [2, 3]
print(nums[2:])          # from index 2 to the end
print(nums[:2])          # from the start up to (not including) index 2
print(nums[::-1])        # a reversed copy (step of -1)
print(sorted(nums))
nums[2:4] = [8, 9]       # assign to a slice
print(nums)
````

## Exercise 5: Find total number of **_unique_** skills for all apps.
Hint: use a Non-primitive datatype which naturally allows this feature

````python
# 5. TYPE YOUR ANSWER/CODE HERE
# "Skills" = the programming languages / tools listed in the free-text `languages` answer.
# The answers are messy ("Python, C/C++", "Python java", "Phyton", "python，R"), so we
# normalize separators, fix a few spellings, and keep only recognised language/tool names.
KNOWN_SKILLS = {
    "python", "java", "c", "c++", "r", "sql", "javascript", "typescript", "html", "css",
    "go", "matlab", "ruby", "haskell", "swift", "bash", "lua", "php", "abap", "stata",
    "excel", "vue",
}
ALIASES = {"phyton": "python", "golang": "go", "js": "javascript", "c/c++": "c c++"}

skills = set()                      # a set keeps only unique values
for app in apps:
    text = app["languages"].lower()
    for sep in [",", "，", ";", "(", ")", "\n", "."]:
        text = text.replace(sep, " ")
    for word in text.split():
        for token in ALIASES.get(word, word).split("/"):
            token = ALIASES.get(token, token)
            for part in token.split():
                if part in KNOWN_SKILLS:
                    skills.add(part)

print("Unique skills:", len(skills))
print(sorted(skills))
````

> **How many skipped optional questions?**

Truthiness gives us a concise test for blank answers:

````python
skipped_languages = 0
skipped_additional = 0

for app in apps:
    if not app["languages"]:          # an empty string is falsy, so `not` makes the test true
        skipped_languages += 1
    if not app["additional"]:
        skipped_additional += 1

print("Left 'languages' blank: ", skipped_languages)
print("Left 'additional' blank:", skipped_additional)
````

Loops can be **nested**. Here we audit every column at once:

````python
print(COLUMNS) #given from boilerplate code.
````

````python
for question in COLUMNS:              # iterating over a dictionary yields its keys
    blanks = 0
    for app in apps:
        if app[question].strip() == "":
            blanks += 1
    print(question.rjust(14), "blank answers:", blanks)
````

### `while` loops

A `while` loop repeats its body **as long as** its condition holds. It is the right tool when the number of iterations is not known in advance.

```python

    ...
```

> **Question How many people applied before the final day?**
>


*   The form closed at the end of **9/13**.
*   Rows are stored in the order of their submission timestamp,
*   So we can walk down the list until we reach the first application dated 9/13.
* We do not know in advance how many steps that will take, which is precisely the situation `while` is designed for.

````python
i = 0
while not apps[i]["timestamp"].startswith("9/13"):
    i += 1

print("The first 9/13 application is at index", i)
print("So", i, "people applied before the final day")
````

````python
# Equivalent formulation using break
i = 0
while True:                  # loop indefinitely...
    if apps[i]["timestamp"].startswith("9/13"):
        break                # ...until break exits the loop
    i += 1

print("Same result with break:", i)
before_last_day = i
````

## Exercise 6: A classic: **count the number of distinct words in a text.** Apply it a random applicant's essay, eg: `apps[24]["essay_excited"]`.

*Hint: `.lower()`, `.replace()` to strip punctuation, `.split()`, and a `set`.*

````python
text = apps[24]["essay_excited"]
# 6. TYPE YOUR ANSWER/CODE HERE
def count_distinct_words(text):
    """Return the number of distinct words, ignoring case and punctuation."""
    clean = text.lower()
    for mark in ".,!?;:()\"'-–—/":
        clean = clean.replace(mark, " ")
    return len(set(clean.split()))

print("Total words:   ", len(text.split()))
print("Distinct words:", count_distinct_words(text))
````

**`return` versus `print`.** `print` *displays* a value; `return` *delivers* it to the caller, where it can be stored and reused. A function without an explicit `return` returns `None`.

````python
def double_hours(hours):
    print(hours * 2)

def double_hours_return(hours):
    return hours * 2

a = double_hours(5)      # displays 10, returns None
b = double_hours_return(5)     # displays nothing, returns 10
print("a =", a)
print("b =", b)
````

````python
def time_of_day(hour):
    if 5 <= hour < 12:
        return "morning"
    elif 12 <= hour < 17:
        return "afternoon"
    elif 17 <= hour < 22:
        return "evening"
    else:
        return "night"

for h in [3, 9, 15, 20]:
    print(h, time_of_day(h))
````

The rows from the case study reappear: three different spellings of `5 - 7.5 hours`.

We will write **cleaning functions** and store the cleaned values under **new keys** in each application. Deriving new variables from existing ones is called **feature engineering**.

````python
def hours_to_number(answer):
    """Return the minimum weekly hours as a number: '5 - 7.5 hours' -> 5.0"""
    answer = answer.replace("–", "-")      # en dash -> ordinary hyphen, so every answer splits the same way
    low = answer.split("-")[0]             # '5 - 7.5 hours' -> '5 '
    return float(low)                      # float() ignores the extra spaces


def section_number(answer):
    """Return the section as an int: 'Section 2: Thursdays (1:00 - 3:00 PM ET)' -> 2"""
    label = answer.split(":")[0]           # 'Section 2'
    return int(label.split(" ")[1])        # '2' -> 2


# test functions before trusting them
print(first['hours'])
print(first['section'])

print(hours_to_number(first['hours']))
print(section_number(first['section']))
````

# Part 7 · Scope

---

A variable's **scope** is the region of code in which its name can be used.
1. **Global scope:** names defined at the top level of the notebook can be *read* from within functions.
2. **Local scope:** names created *inside* a function, including its parameters, exist only while that call is executing and are invisible outside it.

The last line of the next cell raises an exception by design.

````python
DEADLINE = "9/13/2026"            # global: visible everywhere

def count_on_deadline(rows):
    on_deadline = 0               # local: exists only inside this function
    for row in rows:
        if row["timestamp"].startswith(DEADLINE):   # reading a global is permitted
            on_deadline += 1
    return on_deadline

print("Submitted on the deadline day:", count_on_deadline(apps))
print("DEADLINE is still:", DEADLINE)
# print(on_deadline)                # local names do not survive the call
````

````python
def add_bonus_hours(hours):
    hours = hours + 2             # rebinds the LOCAL name only
    print("Inside the function:", hours)
    return hours

my_hours = 5
print("Returned value:", add_bonus_hours(my_hours))
print("my_hours outside is unchanged:", my_hours)
````

**Mutable arguments behave differently.** Recall aliasing from Part 5: passing a list or dictionary to a function passes a reference to the *same* object. If the function *mutates* that object (for example with `.append()`), the change is visible to the caller. This is exactly how our feature-engineering loops added new keys to every application.

````python
def add_language(lang_list, lang):
    lang_list.append(lang)        # mutates the caller's list

my_langs = ["Python"]
add_language(my_langs, "SQL")
print(my_langs)                   # ['Python', 'SQL']
````
